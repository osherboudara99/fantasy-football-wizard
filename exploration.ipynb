```python
# Remove nulls
df = df[~(df['player_display_name'].isnull()) & ~(df['position'].isnull())]
df.isnull().sum()
```

```python
columns = ['completions', 'attempts',
       'passing_yards', 'passing_tds', 'interceptions', 'sacks', 'sack_yards',
       'sack_fumbles', 'sack_fumbles_lost', 'passing_air_yards',
       'passing_yards_after_catch', 'passing_first_downs', 
       'passing_2pt_conversions', 'carries', 'rushing_yards',
       'rushing_tds', 'rushing_fumbles', 'rushing_fumbles_lost',
       'rushing_first_downs', 'rushing_2pt_conversions',
       'receptions', 'targets', 'receiving_yards', 'receiving_tds',
       'receiving_fumbles', 'receiving_fumbles_lost', 'receiving_air_yards',
       'receiving_yards_after_catch', 'receiving_first_downs',
       'receiving_2pt_conversions', 'special_teams_tds', 'fantasy_points', 'fantasy_points_ppr']

seasonal_agg_df = df.groupby(
    ['player_id', 'player_display_name', 'position', 'position_group', 'season', 'season_type'], as_index=False)[columns].sum()
```
